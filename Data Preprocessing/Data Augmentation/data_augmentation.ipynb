```python
#For working with file paths and creating directories
import os
#For loading and saving medical image files in NIfTI format
import nibabel as nib
#For numerical operations and randomisation
import numpy as np
#For image transformations such as rotation and scaling
from scipy.ndimage import rotate, zoom
```

```python
#Apply a random rotation to a 3D image within the specified angle range - 15
#Rotates around the axial plane (axes 0 and 1), which is standard in neuroimages
#reshape=False preserves original shape; order=1 uses linear interpolation
def random_rotate(data, angle_range=(-15,15)):
    angle = np.random.uniform(angle_range[0], angle_range[1])
    return rotate(data, angle, axes=(0,1), reshape=False, order=1)
```

```python
#Randomly flip the 3D image along the sagittal axis (axis 0)
#Controlled by a flip probability (default is 0.5)
def random_flip(data, flip_prob=0.5):
    if np.random.rand() < flip_prob:
        return np.flip(data, axis=0)
    return data
```

```python
#Apply random uniform scaling to the 3D image within a given scale range
#The same factor is applied to all three spatial dimensions
def random_scale(data, scale_range=(0.9,1.1)):
    scale = np.random.uniform(scale_range[0], scale_range[1])
    factors = [scale, scale, scale]
    return zoom(data, factors, order=1)
```

```python
#Create output directory for augmented Alzheimer's disease (AD) PET scans
os.makedirs('ad_augmented', exist_ok=True)
#Loop through each AD PET scan, apply random transformations, and save output
for i in range(1, 73):
    ad_path = f'ad/registered_pet_ad_{i}.nii.gz'
    img = nib.load(ad_path)
    
    #Apply random augmentations to increase dataset variability
    data = img.get_fdata()
    data = random_rotate(data)
    data = random_flip(data)
    data = random_scale(data)
    
    #Save the augmented scan
    nib.save(nib.Nifti1Image(data, img.affine), f'ad_augmented/AD_Augmented{i}.nii.gz')
```

```python
#Create output directory for augmented cognitively normal (CN) PET scans
os.makedirs('cn_augmented', exist_ok=True)
#Loop through each CN PET scan, apply the same augmentations, and save output
for i in range(1, 100):
    cn_path = f'cn/registered_pet_cn_{i}.nii.gz'
    img = nib.load(cn_path)
    data = img.get_fdata()
    
    #Apply random augmentations
    data = random_rotate(data)
    data = random_flip(data)
    data = random_scale(data)
    
    #Save the augmented scan
    nib.save(nib.Nifti1Image(data, img.affine), f'cn_augmented/CN_Augmentation{i}.nii.gz')
```
